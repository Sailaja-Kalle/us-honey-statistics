**Top Honey Producing State**

```python
df2 = df.sort_values("production",ascending=False)
```

```python
df2
```

```python
plt.figure(figsize=(10,10))
sns.barplot(x=df2["state"],y=df2["production"],ci=None)
plt.xlabel("Honey Producing states")
plt.ylabel("Total Honey Production")
plt.xticks(rotation=90)
plt.show()
```

```python
# from the above graph we can conclude that the top 5 honey producing states are listed below:
#     North Dakota
#     Clifornia
#     South Dakota
#     Florida
#     Montana
```

**Change in mean Average price of Honey from 1995 to 2021**

```python
df1 = df.groupby("year").mean()
```

```python
df1
```

```python
df1.reset_index(inplace=True)
```

```python
df1
```

```python
plt.figure(figsize=(15,5))
sns.barplot(x=df1["year"],y=df1["average_price"])

for index, value in enumerate(df1["average_price"]):
    plt.annotate('{:.2f}'.format(value), xy=(index, value), ha='center', va='bottom')
    
plt.xlabel("Year(1995-2021)")
plt.ylabel("Average price of Honey")
plt.show()
```

**Which was the year when total mean value of production of Honey was the highest ?**

```python
df2 = df2.groupby(["year"]).mean()
```

```python
df2
```

```python
df2.reset_index(inplace=True)
```

```python
df2
```

```python
plt.figure(figsize=(15,5))
sns.barplot(x=df2["year"],y=df2["value_of_production"])
plt.xlabel("Year(1995-2021)")
plt.ylabel("Value of Production of Honey")
plt.show()
```

```python
plt.figure(figsize=(15,5))
sns.scatterplot(x=df2["year"],y=df2["value_of_production"])
plt.xlabel("Year(1995-2021)")
plt.ylabel("Value of Production of Honey")
plt.show()
```

```python
plt.figure(figsize=(10,10))
plt.pie(df2["value_of_production"],labels=df2["year"],autopct="%1.4f%%")
plt.show()
```

from the above graphs we can infer that the mean value of production of honey was maximum in year 2014
